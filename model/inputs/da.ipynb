```python
import pandas as pd
loads = pd.read_csv('./loads.csv')
timepoints = pd.read_csv('./timepoints.csv')

loads = pd.merge(timepoints, loads, left_on='timepoint_id', right_on='timepoints', how="left")
loads = loads[['LOAD_ZONE','timepoints','zone_demand_mw']]
loads.to_csv('./del.csv', index=False)
loads
```

```python
variable_capacity_factors = pd.read_csv('./variable_capacity_factors.csv')
timepoints = pd.read_csv('./timepoints.csv')

variable_capacity_factors = pd.merge(variable_capacity_factors, timepoints, on='timepoint_id', how="right")
variable_capacity_factors = variable_capacity_factors[['GENERATION_PROJECT','timepoint_id','gen_max_capacity_factor']]
variable_capacity_factors.to_csv('./del.csv', index=False)
variable_capacity_factors
```
